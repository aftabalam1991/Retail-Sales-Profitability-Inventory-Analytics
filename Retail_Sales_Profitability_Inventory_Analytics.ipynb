# Strong Analysis: Break-even Discount 

#### Break-even discount % = (1 − Unit Cost / MRP) × 100

#### Add product cost details 

```python
product_analysis = product_analysis.merge(
    products[
        ["Product_ID", "Brand", "MRP", "Unit_Cost"]
    ],
    on="Product_ID",
    how="left"
)
```

#### Calculate Break-even Discount 

```python
product_analysis["Break_Even_Discount_Pct"] = (
    (
        1 -
        product_analysis["Unit_Cost"]
        / product_analysis["MRP"]
    ) * 100
).round(2)
```

```python
product_analysis[[
    "Product_ID",
    "Product_Name",
    "Category",
    "MRP",
    "Unit_Cost",
    "Gross_Margin_Pct",
    "Break_Even_Discount_Pct",
    "Loss_Transaction_Rate_Pct",
    "Gross_Loss"
]].sort_values(
    "Loss_Transaction_Rate_Pct",
    ascending=False
).head(15)
```

#### Products will now be classified into risk bands:

```python
import numpy as np

product_analysis["Discount_Risk"] = np.where(
    product_analysis["Break_Even_Discount_Pct"] < 40,
    "High Risk",
    "Normal Risk"
)
```

```python
product_analysis["Discount_Risk"].value_counts()
```

```python
# High Risk products

high_risk_products = (
    product_analysis[
        product_analysis["Discount_Risk"] == "High Risk"
    ]
    .sort_values("Break_Even_Discount_Pct")
)

high_risk_products[[
    "Product_ID",
    "Product_Name",
    "Category",
    "MRP",
    "Unit_Cost",
    "Break_Even_Discount_Pct",
    "Loss_Transaction_Rate_Pct",
    "Gross_Loss"
]]
```

```python
# Just "High Risk" / "Normal Risk" is a bit basic. Create three risk bands using Numpy.

conditions = [
    product_analysis["Break_Even_Discount_Pct"] < 35,
    product_analysis["Break_Even_Discount_Pct"] < 40
]

choices = [
    "Critical Risk",
    "High Risk"
]

product_analysis["Discount_Risk_Band"] = np.select(
    conditions,
    choices,
    default="Normal Risk"
)
```

```python
product_analysis["Discount_Risk_Band"].value_counts()
```

```python
product_analysis[
    [
        "Product_ID",
        "Product_Name",
        "Break_Even_Discount_Pct",
        "Discount_Risk_Band"
    ]
].sort_values("Break_Even_Discount_Pct").head(15)
```

### Now, we will validate the rule against actual transactions.

```python
# First, add the break-even value to the transaction data.

break_even_lookup = product_analysis[
    ["Product_ID", "Break_Even_Discount_Pct"]
]

discount_40 = (
    sales_analysis[sales_analysis["Discount_Pct"] == 40]
    .merge(
        break_even_lookup,
        on="Product_ID",
        how="left"
    )
)
```

```python
# Now, theoretically, should a loss occur or not?

discount_40["Expected_Loss"] = (
    discount_40["Discount_Pct"]
    >
    discount_40["Break_Even_Discount_Pct"]
)
```

```python
# Actual transaction loss

discount_40["Actual_Loss"] = (
    discount_40["Gross_Profit"] < 0
)
```

```python
# Comparision

pd.crosstab(
    discount_40["Expected_Loss"],
    discount_40["Actual_Loss"]
)
```

### Management Recommendation

### Recommended discount cap 

```python
product_analysis["Recommended_Max_Discount_Pct"] = (
    np.floor(
        (product_analysis["Break_Even_Discount_Pct"] - 5) / 5
    ) * 5
).clip(lower=0)
```

```python
company_margin_2025 = (
    sales_2025["Gross_Profit"].sum()
    / sales_2025["Net_Sales"].sum()
    * 100
)

company_atv_2025 = (
    sales_2025["Net_Sales"].sum()
    / sales_2025["Transaction_ID"].nunique()
)

company_upt_2025 = (
    sales_2025["Quantity"].sum()
    / sales_2025["Transaction_ID"].nunique()
)

print(f"2025 Company Margin: {company_margin_2025:.2f}%")
print(f"2025 Company ATV: ₹{company_atv_2025:,.2f}")
print(f"2025 Company UPT: {company_upt_2025:.2f}")
```
